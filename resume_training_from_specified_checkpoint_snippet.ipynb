# demo1 從指定的目錄下選最新的checkpoint, 再繼續training

```python
# SW_resume_training = False   # 重新training
SW_resume_training = True    # resume training from a specified checkpoint(.PTH)
```

# demo2 從指定的目錄下選最好的checkpoint(best_model), 再繼續training

```python
#指定的目錄下，直接指定checkpoint （如 checkpoint_6482.pth 或 best_model.pth）
import os

# 指定module目錄
OUTPUT_MODEL_PATH = "/content/drive/MyDrive/Colab_Notebooks/train/"
TRAIN_MOULE_SUBDIR = "train-October-21-2025_03+23PM-0000000"

#指向要載入的 .pth 檔案（如 checkpoint_6482.pth 或 best_model.pth）。
RESUME_MODEL_PATH = os.path.join(OUTPUT_MODEL_PATH, TRAIN_MOULE_SUBDIR, 'best_model.pth' )
print(RESUME_MODEL_PATH)
```
